# The Mojo Programming Language

MMMAudio leverages the Mojo programming language to create c++-level performance DSP code in a pythonic syntax. If you're coming with experience from another language like Python, Javascript, or sclang there are a few important differences to understand with Mojo:

1. Mojo is a __statically typed__ language. All variable and argument declaration *require* a type definition.
2. Mojo code is __compiled__ (meaning you can't make changes to it at runtime)
3. Mojo uses *structs* and *Traits* for polymorphism instead of *classes*. There is no inheritance in Mojo. 

## Mojo Data Types

Mojo has a number of data types to know. The most commonly used in MMMAudio are:

* ```Float64``` - 64bit floats
* ```Int``` - 32bit or 64bit signed integers
* ```String``` - A collection of characters
* ```Bool``` - True or False

All variable declarations require a type designation. For instance, we may want to have a variable that holds the frequency of an oscillator. To do this we would write the following code:

```mojo
var freq : Float64 = 440.0
```

This is the safest way to do this. However, Mojo can (situationally) interpret a variable's data type at *compile time*. The following would also be valid code:

```mojo
var freq = 440.0 # Interpreted as a Float64 at compile time
```

Note that a decimal place is included in this declaration. If the decimal place was ommitted the variable would be interpreted as an Int at compile time.

```mojo
var feq = 440 # Interpreted as an Int at compile time
```

### Mojo Data Structures

Mojo has a 3 data structures that are commonly used with MMMAudio:

* Lists - A resizeable collection of the *same* data type
* Arrays - A fixed size collection of the *same* data type
* SIMD Vectors - A Single Input Multiple Direction Vector of a single *data* type. Must be a power of two in size.

All of these data structures are *typed*. Below are examples of creating a data structure that holds ```FLoat64```s:

```mojo
 # We can put as many floats in this list as we would like.
var my_float_list : List[Float64] = [1.0, 2.0, 3.0]


# This creates an array of length 3 with 0.0 at each index. We cannot increase the size of this array after its declaration.
var my_float_array : Array[Float64, 3] = Array[Float64, 3](fill=0.0) 


# This creates a SIMD vector of length four with 0.0 at each index. MFloat is a convenience struct included with MMMAudio. A SIMD vector must be a power of two in length. 
var my_float_simd : MFloat[4] = MFloat[4](0.0)
```
