# The Mojo Programming Language

MMMAudio leverages the Mojo programming language to create c++-level performance DSP code in a pythonic syntax. If you're coming with experience from another language like Python, Javascript, or sclang there are a few important differences to know about Mojo:

1. Mojo is a __statically typed__ language. All variable and argument declaration *require* a type definition.
2. Mojo code is __compiled__ (meaning you can't make changes to it at runtime)
3. Mojo uses *structs* and *Traits* for polymorphism instead of *classes*. There is no inheritance in Mojo. 

## Mojo Data Types

Mojo has a number of data types to know. The most commonly used in MMMAudio are:

* ```Float64``` - 64bit floats
* ```Int``` - 32bit or 64bit signed integers
* ```String``` - A collection of characters
* ```Bool``` - True or False

All variable declarations require a type designation. For instance, we may want to have a variable that holds the frequency of an oscillator. To do this we would write the following code:

```mojo
var freq : Float64 = 440.0
```

This is the safest way to do this. However, Mojo can (situationally) interpret a variable's data type at *compile time*. The following would also be valid code:

```mojo
var freq = 440.0 # Interpreted as a Float64 at compile time
```

Note that a decimal place is included in this declaration. If the decimal place was ommitted the variable would be interpreted as an Int at compile time.

```mojo
var feq = 440 # Interpreted as an Int at compile time
```

## Mojo Data Structures

Mojo has a 3 data structures that are commonly used with MMMAudio:

* Lists - A resizeable collection of the *same* data type
* Arrays - A fixed size collection of the *same* data type
* SIMD Vectors - A Single Input Multiple Direction Vector of a single *data* type. Must be a power of two in size.

All of these data structures are *typed*. Below are examples of creating a data structure that holds ```FLoat64```s:

```mojo
 # We can put as many floats in this list as we would like.
var my_float_list : List[Float64] = [1.0, 2.0, 3.0]


# This creates an array of length 3 with 0.0 at each index. We cannot increase the size of this array after its declaration.
var my_float_array : Array[Float64, 3] = Array[Float64, 3](fill=0.0) 


# This creates a SIMD vector of length four with 0.0 at each index. MFloat is a convenience struct included with MMMAudio. A SIMD vector must be a power of two in length. 
var my_float_simd : MFloat[4] = MFloat[4](0.0)
```



## Compile time vs. Run time

Mojo allows for the explicit definition of operations to run at *compile time* and *run time*. Compile time operations occur when a .mojo file is *compiled*, happening only once. This necessarily means that a compile time calculation will be static through the program's lifespan. In contrast, a *run time* operation is calculated whenever the relevant portion of the program is run. *Run time* calculations are dynamic and can produce different results while the program is running. Take for example a variable that stores the value 2pi in it. Here are both a run time declaration and an explicit compile time declaration:

```mojo
# Run time
var two_pi = 2 * pi
```

```mojo
# Compile time
comptime two_pi = 2 * pi
```

The compiletime keyword explicitly tells the mojo compiler to calculate the variable at compile time instead of at run time, potentially stopping needless calculations on a static value. Mojo also allows for the use of *compile time arguments* called *parameters* for functions and ```structs```. Here are two equivalent functions, one with a run time argument and the other a compile time parameter:

```mojo
# Run time
def is_between_speakers(speaker_positions: Array[Float64, 8], speaker_a: Int, speaker_b: Int, input_az: Float64) -> Bool:
    if input_az > speaker_positions[speaker_a] and input_az < speaker_positons[speaker_b]:
        return True
    else:
        return False
```

```mojo
# Compile time
def is_between_speakers[speaker_positions: Array[Float64, 8]](speaker_a: Int, speaker_b: Int, input_az: Float64) -> Bool:
    if input_az > speaker_positions[speaker_a] and input_az < speaker_positons[speaker_b]:
        return True
    else:
        return False
```

## Object Oriented Programming

### Creating a Class/Struct

#### Mojo

```mojo

struct StructName(StructTraits):
    var float_property: Float64
    var string_property: String

    def __init__(float_arg: Float64 = 1.0, string_arg: String = "Gondor calls for aid"):
        #constructor
        self.float_property = float_arg
        self.string_property = string_arg

    def instance_method(method_arg: Float64) -> Float64:
        var output = self.float_property + method_arg
        return output
```

#### Python

````python
class ClassName(ClassInheritedFrom):
    
    def __init__(float_arg = 1.0, string_arg = "Gondor calls for aid"):
        #constructor
        self.float_property = float_arg
        self.string_property = string_arg
        
    def instance_method(method_arg):
        output = self.float_property + method_arg
        return output
````

#### SClang

```sclang

    var <>string_property;

    *new {

		arg float_arg = 1.0, string_arg = "Gondor calls for aid";
        float_property = float_arg;
        string_property = string_arg;
    }

    instance_method {
        arg method_arg;
        var output;

    }
}
```
